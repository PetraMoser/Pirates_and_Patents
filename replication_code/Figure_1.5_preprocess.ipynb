```python
# Import the necessary packages
import pandas as pd
import csv
import os
from datasets import load_dataset
```

```python
# Load dataset directly from Hugging Face
dataset_all_years = load_dataset(
    "matthewleechen/300YearsOfBritishPatents",
    data_files="all_patents.jsonl.gz"
)

# Convert to a readable data frame and aggregate number of patente per year
df = (
    dataset_all_years['train']
        .to_pandas()[['year']]        # only load the year column
        .groupby('year')
        .size()
        .reset_index(name='patent_count'))

df.head()
```

```python
df.to_csv('../Data/patents_per_year.csv', index=False)
```
